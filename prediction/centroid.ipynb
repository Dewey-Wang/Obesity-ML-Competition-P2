```python
import scanpy as sc
import pandas as pd
import numpy as np
import os

# read data
adata = sc.read_h5ad("../data/obesity_challenge_2.h5ad")

genes = adata.obs["gene"].astype(str)

# masks
single_mask = (~genes.str.contains("\+")) & (genes != "NC")

double_mask = (genes.str.count("\+") == 1) & (genes != "NC+NC")

# extract matrices
X_single = adata[single_mask].X
X_double = adata[double_mask].X

# convert sparse -> dense if needed
if not isinstance(X_single, np.ndarray):
    X_single = X_single.toarray()

if not isinstance(X_double, np.ndarray):
    X_double = X_double.toarray()

# compute centroids
single_centroid = X_single.mean(axis=0)
double_centroid = X_double.mean(axis=0)

# gene names = columns
gene_names = adata.var_names.astype(str)

# create dataframe
centroid_df = pd.DataFrame(
    [single_centroid, double_centroid],
    index=["single_centroid", "double_centroid"],
    columns=gene_names
)

# ensure directory exists
output_dir = "../data/preprocessed"
os.makedirs(output_dir, exist_ok=True)

# save
output_path = os.path.join(output_dir, "centroid.csv")
centroid_df.to_csv(output_path)

print(f"Saved centroid file to: {output_path}")
print(centroid_df.shape)
```
