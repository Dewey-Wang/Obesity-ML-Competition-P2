```python
import scanpy as sc
import pandas as pd
import numpy as np

# load data
adata = sc.read_h5ad("../data/obesity_challenge_2.h5ad")

# 確保欄位存在
required_cols = ["gene", "pre_adipo", "adipo", "lipo", "other"]

for c in required_cols:
    assert c in adata.obs.columns, f"missing column {c}"

# 所有 perturbations
genes = ['STAT5A']#adata.obs["gene"].unique()

results = []

eps = 1e-20

for g in genes:

    subset = adata[adata.obs["gene"] == g]

    pre_adipo = subset.obs["pre_adipo"].mean()
    adipo     = subset.obs["adipo"].mean()
    lipo      = subset.obs["lipo"].mean()
    other     = subset.obs["other"].mean()

    lipo_adipo = lipo / (adipo + eps)

    results.append({
        "gene": g,
        "pre_adipo": pre_adipo,
        "adipo": adipo,
        "other": other,
        "lipo": lipo,
        "lipo_adipo": lipo_adipo
    })

df = pd.DataFrame(results)

# optional: 排序
df = df.sort_values("gene").reset_index(drop=True)

print(df.head())
```
